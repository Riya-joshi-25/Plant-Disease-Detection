```python
# Create the path of the main images folder and This folder contains separate folders for different plant diseases
image_path=os.path.join(r'c:\\Users\\RIya\\Desktop\\6month-Data-Science\\deeplearning\\CNN\\Plant-Disease-Detection',"Data",'images')


# Display all folders/classes present inside the images folder and Each folder represents a different disease/class 
os.listdir(image_path)
```

```python
# Create an empty dictionary to store image paths and Each key represents a disease/class
# The corresponding value will contain paths of images
path_dict={
    "Apple_Black_rot":[],
    "Apple_Cedar_apple_rust":[]
}

# Loop through every folder inside the images directory
for key in os.listdir(image_path):


    # Create the complete path of the current disease folder
    dir_path=os.path.join(image_path,key)

     # Loop through every image inside the current disease folder
    for image in os.listdir(dir_path):

        # Create the complete path of the image and  Then store that path inside the dictionary
        path_dict[key].append(os.path.join(dir_path,image))

```
